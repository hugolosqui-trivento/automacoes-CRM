```python

url = 'https://docs.google.com/spreadsheets/d/13DBRdUsHT5wupKkq3Omi9VUr9Nj3R5gYVMfSSkmtXQM/export?format=xlsx'

matriculas_itabirito = pd.read_excel(url, sheet_name = "ITABIRITO")
matriculas_altamira = pd.read_excel(url, sheet_name = "ALTAMIRA")
matriculas_lorena = pd.read_excel(url, sheet_name = "LORENA")
matriculas_canaa = pd.read_excel(url, sheet_name = "Canaã")


matriculas = pd.concat([matriculas_itabirito, matriculas_altamira, matriculas_lorena, matriculas_canaa], ignore_index=True)
```

```python
matriculas.shape
```

```python
matriculas.dropna(inplace = True, subset = ['Número de Inscrição'])
```

```python
colunas_captacao = ['Nome Completo_x', 'cpf', 'ID Lead', 'Proprietário', 'Situação Inscrição', 'Número Inscrição', 'Unidade Lead', 'Unidade Form', 'Oferta Atual']
captacao = captacao[colunas_captacao]
```

```python
# reitando Medicina 
captacao = captacao.loc[captacao['Oferta Atual'] != 'Medicina', :]
```

```python
# retirando cursos livres

captacao = captacao.loc[captacao['Unidade Form'].str.contains('Cursos Livres') == False, :]
```

# Mapeando Matriculas entre as tabelas

```python
matriculas_captacao = captacao.loc[captacao['Situação Inscrição'] == 'Matriculado']
```

```python
matriculas_captacao.shape[0]
```

## Matriculas que estão no CRM, mas não estão na planilha

```python
df = matriculas_captacao.loc[~matriculas_captacao['Número Inscrição'].isin(matriculas['Número de Inscrição'].str.strip())]
display(df)
df.shape
```

```python
df.to_excel('matriculas_na_captacao_que_nao_estao_na_planilha.xlsx', index=False)
```

```python
df = matriculas.loc[~matriculas['Número de Inscrição'].isin(matriculas_captacao['Número Inscrição'])]
# eliminar linhas em branco
df = df.dropna(subset=['Número de Inscrição'])
display(df)
df.shape
```
